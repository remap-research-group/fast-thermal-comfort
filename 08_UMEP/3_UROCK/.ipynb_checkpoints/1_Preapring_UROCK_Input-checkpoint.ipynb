```python
from osgeo import gdal

# Input GeoJSON file
input_geojson = "/path/to/input_file.geojson"

# Output Shapefile
output_shp = "/path/to/output_file.shp"

# Convert GeoJSON to Shapefile
gdal.VectorTranslate(output_shp, input_geojson, format="ESRI Shapefile")

print(f"Conversion completed: {output_shp}")
```

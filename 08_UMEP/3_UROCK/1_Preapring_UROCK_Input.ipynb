# Convert GeoJSON into Shapefile

UROCK requires two inputs: a shapefile and a reference GeoTIFF. The GeoTIFF defines the output's extent and coordinate system, but does not otherwise affect the output.
This code uses gdal.VectorTranslate to convert .geojson to .shp (ESRI Shapefile).


```python
from osgeo import gdal

# Input GeoJSON file
input_geojson = "/path/to/input_file.geojson"

# Output Shapefile
output_shp = "/path/to/output_file.shp"

# Convert GeoJSON to Shapefile
gdal.VectorTranslate(output_shp, input_geojson, format="ESRI Shapefile")

print(f"Conversion completed: {output_shp}")
```

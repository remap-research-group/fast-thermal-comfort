```python
# Fixing LULC


import rasterio
import numpy as np

input_path = "/storage/project/r-rbasu31-0/shared/Metro_Boston/01_Input/LULC_5class.tif"
output_path = "/storage/project/r-rbasu31-0/shared/Metro_Boston/01_Input/LULC_5class_fixed.tif"

with rasterio.open(input_path) as src:
    data = src.read()
    profile = src.profile.copy()

    # If the file had a previous nodata value, replace those pixels with 7
    old_nodata = src.nodata
    if old_nodata is not None:
        data = np.where(data == old_nodata, 7, data)

    # Update profile to set new nodata value
    profile.update(nodata=7)

    with rasterio.open(output_path, 'w', **profile) as dst:
        dst.write(data)

print(f"Saved {output_path} with nodata=7")
```
